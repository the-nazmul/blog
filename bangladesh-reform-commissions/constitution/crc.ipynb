```python
import json
from collections import defaultdict
import os

```

```python

# --- CONFIG ---
input_file = "crc.json"            # Your full JSON file
output_dir = "split_parts"         # Folder where split files will be saved

# --- LOAD DATA ---
with open(input_file, 'r', encoding='utf-8') as f:
    data = json.load(f)

# --- GROUP BY "Part" ---
parts_map = defaultdict(list)
for row in data:
    part = row.get("Part", "[অনির্ধারিত অধ্যায়]").strip()
    parts_map[part].append(row)

# --- PREP OUTPUT DIR ---
os.makedirs(output_dir, exist_ok=True)

# --- WRITE parts.json (list of unique part names) ---
part_keys = list(parts_map.keys())
with open(os.path.join(output_dir, 'parts.json'), 'w', encoding='utf-8') as f:
    json.dump(part_keys, f, ensure_ascii=False, indent=2)

# --- WRITE part-0.json, part-1.json, ... ---
for i, part in enumerate(part_keys):
    part_data = parts_map[part]
    filename = f'part-{i}.json'
    with open(os.path.join(output_dir, filename), 'w', encoding='utf-8') as f:
        json.dump(part_data, f, ensure_ascii=False, indent=2)

print(f"✅ Done! {len(part_keys)} part files saved in ./{output_dir}/")

```
